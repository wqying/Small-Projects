```python
def print_catchphrase(character):
    if character == "Pooh":
        print("Oh, bother!")
    elif character == "Tigger":
        print("TTFN: Ta-ta for now!")
    elif character == "Eeyore":
        print("Thanks for noticing me.")
    elif character == "Christopher Robin":
        print("Silly old bear.")
    else:
        print(f"Sorry! I don't know {character}'s catchphrase!")

character = "Maksim"
print_catchphrase(character)
```

```python
def get_item(items: list, x: int):
    if x < len(items):
        return items[x]
    else:
        return None
    
items = ["piglet", "pooh", "roo", "rabbit"]
x = 5
get_item(items, x)
```

```python
def sum_honey(hunny_jars):
    count = 0
    for i in hunny_jars:
        count += i
    return count

hunny_jars = [2, 3, 4, 5]
sum_honey(hunny_jars)
```

```python
def doubled(hunny_jars: list):  # note that the function should return a new list 
    # without modifying the original list
    double_jars = []
    for i in hunny_jars:
        doubled = i * 2
        double_jars.append(doubled)
    return double_jars  # the new list is returned
    
hunny_jars = []
doubled(hunny_jars)
```

```python
def count_less_than(race_times, threshold):
    count = 0
    for i in race_times:
        if i < threshold:
            count += 1
    return count

race_times = [3, 4, 6, 2]
threshold = 3
count_less_than(race_times, threshold)
```

```python
# version 1
def print_todo_list(task):
    print("Pooh's To Dos:")
    count = 0
    for tasks in task:
        count += 1
        print(f"{count}. {tasks}")

# version 2
def print_todo_list(task):
    print("Pooh's To Dos:")
    for tasks in range(1, len(task) + 1):
        print(f"{tasks}. {task[tasks -1]}")

task = ["Count all the bees in the hive", "Chase all the clouds from the sky", "Think", "Stoutness Exercises"]
print_todo_list(task)
```

```python
# dumb implementation
def can_pair(item_quantities):
    pair_count = 0
    for i in item_quantities:
        if i % 2 == 0:
            pair_count += 1
    if pair_count == len(item_quantities):
        return True
    return False
    
# better implementation
def can_pair(item_quantities):
    for i in item_quantities:
        if i % 2 != 0:
            return False
    return True
```
