```python
def print_catchphrase(character):
    if character == "Pooh":
        print("Oh, bother!")
    elif character == "Tigger":
        print("TTFN: Ta-ta for now!")
    elif character == "Eeyore":
        print("Thanks for noticing me.")
    elif character == "Christopher Robin":
        print("Silly old bear.")
    else:
        print(f"Sorry! I don't know {character}'s catchphrase!")

character = "Maksim"
print_catchphrase(character)
```

```python
def count_less_than(race_times, threshold):
    count = 0
    for i in race_times:
        if i < threshold:
            count += 1
    return count

race_times = [3, 4, 6, 2]
threshold = 3
count_less_than(race_times, threshold)
```

```python
# version 1
def print_todo_list(task):
    print("Pooh's To Dos:")
    count = 0
    for tasks in task:
        count += 1
        print(f"{count}. {tasks}")

# version 2
def print_todo_list(task):
    print("Pooh's To Dos:")
    for tasks in range(1, len(task) + 1):
        print(f"{tasks}. {task[tasks -1]}")

task = ["Count all the bees in the hive", "Chase all the clouds from the sky", "Think", "Stoutness Exercises"]
print_todo_list(task)
```

```python
# dumb implementation
def can_pair(item_quantities):
    pair_count = 0
    for i in item_quantities:
        if i % 2 == 0:
            pair_count += 1
    if pair_count == len(item_quantities):
        return True
    return False
    
# better implementation
def can_pair(item_quantities):
    for i in item_quantities:
        if i % 2 != 0:
            return False
    return True
```

```python
def split_haycorns(quantity: int):
    divisors = []
    for i in range(1, quantity + 1):
        if quantity % i == 0:
            divisors.append(i)
    return divisors

quantity = 6
split_haycorns(quantity)
```

```python
def tiggerfy(s: str):
    stolen = "tiger"
    result = ""
    for char in s:
        if char not in stolen:
            result += char
    return result

s = "suspicerous"
tiggerfy(s)
```

```python
def locate_thistles(items):
    word = "thistle"
    indices = []
    for i in range(len(items)):
        if items[i] == word:
            indices.append(i)
    return indices

items = ["thistle", "stick", "carrot", "thistle", "eeyore's tail"]
locate_thistles(items)
```

```python
def are_equivalent(word1, word2):
    word1_list = []
    word2_list = []
    for char1 in word1:
        for chars1 in char1:
            word1_list.append(chars1)
    for char2 in word2:
        for chars2 in char2:
            word2_list.append(chars2)
    if word1_list == word2_list:
        return True
    else:
        return False
    
# Easier implementation:
# use join: join all elements in the data structure with separator
# if dict, joins the keys/items
def are_equivalent2(word1, word2):
    cat_word1 = ''.join(word1)
    cat_word2 = ''.join(word2)
    return cat_word1 == cat_word2
        
```

```python
def count_evens(lst):
    count = 0
    for word in lst:
        if len(word) % 2 == 0:
            count += 1
    return count
```

```python
def remove_name(people, secret_identity):
    for name in people:
        if secret_identity.lower() == name.lower():
            people.remove(secret_identity)
    return people

# Alternative Solution
def remove_name2(people, secret_identity):
    i = 0
    while i < len(people):
        if people[i] == secret_identity:
            people.pop(i)
        else:
            i += 1
    return people

people = ['Batman', 'Superman', 'Bruce Wayne', 'The Riddler', 'Bruce Wayne']
secret_identity = 'Bruce Wayne'
remove_name2(people, secret_identity)
```

```python
def count_digits(n):
    if n == 0:
        return 1
    else:
        i = n // 10
        return len(str(i)) + 1 # cheating solution, bad

# better:
def count_digits2(n):
    if n == 0: # special case
        return 1
    count = 0
    while n > 0:
        n //= 10
        count += 1
    return count
```

```python
def move_zeroes(lst):
    new_list = []
    for num in lst:
        if num == 0:
            lst.remove(num)
            new_list.append(num)
    final = lst + new_list
    return final
    
lst = [1, 0, 2, 0, 3, 0]
move_zeroes(lst)
```

```python
def highest_altitude(gain):
    highest = 0
    current = 0
    for g in gain:
        # don't overthink; anything to do with continuous math is for loop usually
        current += g
        if current > highest:
            highest = current
    return highest

gain = [-4, -3, -2, -1, 4, 3, 2]
highest_altitude(gain)
    
```
